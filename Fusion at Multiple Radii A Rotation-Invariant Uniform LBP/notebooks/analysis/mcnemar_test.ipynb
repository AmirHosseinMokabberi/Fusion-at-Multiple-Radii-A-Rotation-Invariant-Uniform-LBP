```python
import math

def mcnemar_simple(wrong1, wrong2, alpha: float = 0.05, use_cc: bool = True):
    s1, s2 = set(wrong1), set(wrong2)
    b = len(s1 - s2)  # model1 wrong, model2 right
    c = len(s2 - s1)  # model1 right, model2 wrong
    n = b + c

    if n == 0:
        return {"b": b, "c": c, "n": n,
                "chi2": 0.0, "p_value": 1.0, "use_cc": use_cc,
                "significant": False,
                "message": "No discordant pairs; performances are indistinguishable."}

    if use_cc:
        chi2 = (abs(b - c) - 1)**2 / n        # Yates/continuity-corrected
    else:
        chi2 = (b - c)**2 / n                 # uncorrected

    # df=1 -> p = erfc(sqrt(chi2)/sqrt(2))
    p_value = math.erfc(math.sqrt(chi2) / math.sqrt(2.0))
    significant = p_value < alpha

    return {"b": b, "c": c, "n": n, "chi2": chi2, "p_value": p_value,
            "use_cc": use_cc, "significant": significant,
            "message": ("SIGNIFICANT" if significant else "NOT significant")}


# --- Example ---
# model A wrong on [251, 252, 314, 504, 855, 1520, 1589, 1746, 1754]
# model B wrong on the longer list below
if __name__ == "__main__":
    res = mcnemar_simple(
        [26, 27, 97, 152, 207, 221, 248, 250, 251, 252, 314, 350, 373, 374, 425, 431, 504, 765, 853, 854, 855, 940, 999, 1033, 1094, 1200, 1237, 1292, 1310, 1378, 1379, 1380, 1520, 1525, 1589, 1591, 1746, 1754, 1764],
        [251, 252, 314, 504, 855, 1520, 1589, 1746, 1754]
    )
    print(res)
```
